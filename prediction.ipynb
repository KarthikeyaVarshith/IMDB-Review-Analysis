```python
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import load_model
from tensorflow.keras.datasets import imdb
import re
```

```python
model = load_model("imdb_model.h5")
model.summary()
```

```python
model.get_weights()
```

```python
word_index = imdb.get_word_index()
```

```python
def preprocess_txt(txt):
    words = re.sub(r'[^\w\s]', '', txt.lower()).split()

    encoded_review = []

    for word in words:
        if word in word_index:
            encoded_review.append(word_index[word] + 3)
        else:
            encoded_review.append(2)

    padded_review = pad_sequences(
        [encoded_review],
        maxlen=500,
        padding='pre'
    )

    return padded_review
```

```python
def predict_sentiment(review):
    preprocessed_review = preprocess_txt(review)
    prediction = model.predict(preprocessed_review)

    sentiment = 'Positive' if prediction[0][0]>0.5 else 'Negative'
    return sentiment, prediction[0][0] 
```

```python
input_review = "This was an absolutely terrible movie. Don't be fooled by the upbeat packaging or the alluring cover art—the actual film is a complete disaster. The acting is stiff, the dialogue feels entirely forced, and the plot makes absolutely no logical sense whatsoever. I am genuinely shocked that this managed to get funded and distributed. Save your time and your money; skip this completely."
```

```python
predict_sentiment(input_review)
```
